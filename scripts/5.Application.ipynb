```python
from watermark_utils import Watermark_Config, watermark_model
model_name = "google/gemma-3-1b-it"
green_red_list_ratio = 0.5

meaningful_start_at = 237
meaningful_end_at = 255968

wm_config = Watermark_Config(
    watermark_type="v2",
    gamma=4,
    start_at=meaningful_start_at,
    end_at=meaningful_end_at,
    green_red_ratio=green_red_list_ratio,
)
wm_model = watermark_model(model_name, wm_config, device="cpu")
```

```python
messages = [
    [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "**麦茶**を十回言いなさい："},
            ],
        },
    ],
]

response: list[str] = wm_model.generate(
    messages=messages, do_watermark=True, max_new_tokens=256
)
prompt, output = response[0].split("model", maxsplit=1)
print(prompt)
print(output)
print(
    "偏差値 -> ",
    wm_model.detect_watermark(
        output, visualization="../pics/chap5/十回麦茶.svg", title="「麦茶」を十回言いなさい"
    ),
)
```

```python
import torch

def get_unique_random_numbers_torch(start, stop, n, seed=42, device="cpu"):
    """
    startからstopまでの区間の中に、乱数的にn個の整数を選ぶ (Torch版, local seed).
    """
    population_size = stop - start
    if n > population_size:
        raise ValueError(f"区間の長さ{population_size}はn ({n})以下")

    # Local generator to avoid affecting global RNG
    g = torch.Generator(device=device)
    g.manual_seed(seed)

    # torch.randperm uses generator if passed
    result = torch.randperm(population_size, generator=g, device=device)[:n] + start
    return result.sort().values  # sorted tensor
print(get_unique_random_numbers_torch(10, 30, 5, seed=42))
print(get_unique_random_numbers_torch(10, 30, 5, seed=114514))  # same result
```
